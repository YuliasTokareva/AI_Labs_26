```python
import numpy as np

# Генерация тестовых данных строго по примеру из методички
np.random.seed(42)

# X: 100 примеров, 12 бинарных признаков (ответы на вопросы)
X = np.random.randint(0, 2, size=(100, 12)) 

# Y: 100 примеров, 2 класса (one-hot encoding: [1,0] или [0,1])
Y = np.random.randint(0, 2, size=(100, 2))  

# fmt='%d' сохраняет целые числа без лишних нулей после запятой
np.savetxt('dataIn.txt', X, fmt='%d')
np.savetxt('dataOut.txt', Y, fmt='%d')

print("Файлы dataIn.txt и dataOut.txt успешно созданы.")
print(f"Размер входных данных (N примеров x 12 признаков): {X.shape}")
print(f"Размер выходных данных (N примеров x 2 класса): {Y.shape}")
print("\nПример первых 3 строк dataIn.txt:")
print(X[:3])
print("\nПример первых 3 строк dataOut.txt:")
print(Y[:3])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Загрузка данных из файлов
X = np.loadtxt('dataIn.txt')
Y = np.loadtxt('dataOut.txt')

print(f"Загружено {X.shape[0]} примеров с {X.shape[1]} признаками")
print(f"Целевая переменная имеет {Y.shape[1]} класса")

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)

# Нормализация входных данных 
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Создание нейросети 
# Один скрытый слой + функция активации logsig (в Keras это 'sigmoid')
model = keras.Sequential([
    keras.layers.Dense(16, activation='sigmoid', input_shape=(12,)), # Скрытый слой с logsig
    keras.layers.Dense(2, activation='softmax') # Выходной слой для 2 классов
])

# Компиляция модели
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("\nАрхитектура модели:")
model.summary()
```

```python
# Задание 3: Обучение модели на предоставленных данных
# epochs=50 - количество полных проходов по обучающей выборке
# batch_size=16 - размер мини-пакета для стохастического градиентного спуска
# validation_data - позволяет отслеживать переобучение в реальном времени
history = model.fit(
    X_train, y_train, 
    epochs=50, 
    batch_size=16, 
    validation_data=(X_test, y_test),
    verbose=1 # Выводит прогресс-бар обучения
)

print("\nОбучение завершено!")
print(f"Финальная точность на обучении: {history.history['accuracy'][-1]:.4f}")
print(f"Финальная точность на тесте:    {history.history['val_accuracy'][-1]:.4f}")
```
